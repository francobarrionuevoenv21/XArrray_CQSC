# ¿Qué es Xarray?

Xarray es un proyecto de código abierto y un paquete de Python que extiende las capacidades de los arreglos de NumPy, ofreciendo una forma más sencilla e intuitiva de trabajar con datos multidimensionales. Al igual que Pandas facilita el análisis de datos tabulares, Xarray está orientado al manejo de arreglos de N dimensiones, como los que se encuentran habitualmente en aplicaciones climáticas, meteorológicas y ambientales. Su interfaz se inspira en el modelo de datos de netCDF, basado en variables, dimensiones y atributos, pero incorpora herramientas avanzadas para el análisis y procesamiento de datos multidimensionales, similares a las del [Common Data Model (CDM) desarrollado para netCDF-Java](https://docs.unidata.ucar.edu/netcdf-java/current/userguide/common_data_model_overview.html).

````python
# Librerías
import numpy as np
import pandas as pd
import xarray as xr # Importamos xarray como "xr"
import matplotlib.pyplot as plt
import random
````

# Objetos DataArray

```DataArray``` es la estructura de datos básica de Xarray. Puede considerarse como una versión mejorada de un numpy.ndarray, ya que además de almacenar los valores numéricos, conserva información importante sobre ellos, como sus coordenadas y metadatos. Esto aporta dos ventajas principales:

- Las coordenadas (por ejemplo, tiempo, latitud y longitud) se almacenan junto con los datos, lo que permite realizar selecciones e indexaciones de forma más intuitiva y precisa.
- Dispone de un espacio para almacenar atributos o metadatos, como las unidades de medida, una descripción de la variable o cualquier otra información relevante.

## Creación de un DataArray desde una array numpy 3D

### Generación de los datos

````python
# Creamos un array 3D con valores aleatorios de temperatura en K

rand_data = 283 + 5 * np.random.randn(5, 3, 4)

print(f'Dimensiones array: {rand_data.shape}')

print('Array T:')
rand_data
````

### Creación del *xr.DataArray*

DataArray proporciona una capa envolvente para los *ndarrays* de NumPy que utiliza dimensiones y coordenadas etiquetadas para permitir operaciones que tienen en cuenta los metadatos

````python
# Utilizamos los datos previamente generados como input del DataArray
temp_da  = xr.DataArray(rand_data)

print('Visualizamos el DataArray:')
temp_da
````

**Dos detalles a destacar:**

- Si no se especifican nombres para las dimensiones, Xarray les asigna automáticamente nombres genéricos como *dim_0*, *dim_1* y *dim_2*. Si trabajamos con datos ambientales o climáticos, lo lógico será asignarles los nombres de time, latitude y longitude.
- Al convertir un arreglo de NumPy en un DataArray, el notebook muestra una representación mucho más completa del objeto, incluyendo sus dimensiones, coordenadas y atributos. Al hacer clic en el ícono del arreglo, puedes expandir o contraer esta información.

### Creación del DataArray con dimensiones asignadas


Buena parte del poder de usar Xarray se basa en la posibilidad de asignarles nombres a las dimensiones. A continuación, agreguemos una lista ordenada con los nombres de las dimensiones. 

````python
temp_da = xr.DataArray(rand_data, dims=["time", "lat", "lon"])

print("Visualizamos el DataArray con las dimensiones asignadas")
temp_da
````

### Creación de un DataArray con coordenadas

````python
# Creamos datos coordenada time con pandas
times_index = pd.date_range("2018-01-01", periods=5)
times_index
````

````python
# Genrración de los datos cordenadas longitud y latitud
lons = np.linspace(-120, -60, 4)
lats = np.linspace(25, 55, 3)

# Printeamos las dimensiones
print(lons.shape, lats.shape)
````

````python
# Creación del DataArrat
temp_da_coord = xr.DataArray(
    rand_data, coords=[times_index, lats, lons], dims=["time", "lat", "lon"]
)

# Visualización del DataArray
temp_da_coord
````

### Configuración de los atributos

````python
# Defino unidades y nombre de la variable
temp_da_coord.attrs["units"] = "Kelvin"
temp_da_coord.attrs["standard_name"] = "air_temperature"

# Imprimo el DataArray actualizado
temp_da_coord
````

**¡Importante! Incomptabilidades en los atributos**

Al realizar operaciones matemáticas sobre un DataArray, los atributos se preservan. Esto puede dar lugar a errores en quien procese posteriormente los datos. Si hacemos un pasaje de grados K a °C, el atributo de la unidad se mantiene.  

````python
# Pasaje de los datos de grados K a °C
temp_da_celsius = temp_da_coord- 273.15
temp_da_celsius
````

````python
# Actualizamos los atributos
temp_da_celsius.attrs["units"] = "Celsius"

# Verifivamos el DataArray
temp_da_celsius
````

# Objetos Dataset

Junto con ```DataArray```, el otro tipo de objeto fundamental de Xarray es ```Dataset```. Un Dataset es un contenedor similar a un diccionario que almacena uno o más objetos DataArray, los cuales pueden compartir opcionalmente las mismas coordenadas.

## Creación de un Dataset a partir dos DataArrays

### Creación de un DataArray de datos de presión atmosférica

````python
# Repetimos todos los pasos que hicimos para temperatura, ahora para la variable presión
press_data = 1000.0 + 5 * np.random.randn(5, 3, 4)
press_da = xr.DataArray(
    press_data, coords=[times_index, lats, lons], dims=["time", "lat", "lon"]
)
press_da.attrs["units"] = "hPa"
press_da.attrs["standard_name"] = "air_pressure"

# Visualizamos el DataArray creado
press_da
````

### Creación del Dataset

Al momento de crear el objeto Dataset a partir de DataArrays, hay dos cuestiones importantes a tener en cuenta

- La forma más directa de crear el Dataset es pasando como input del argumento ```data_vars``` un diccionario que tome como palabra *clave (Key)* el nombre de la variable, y como *valor (Value)* el DataArray con los datos. 
- El objeto Dataset adopta las coordenadas del último DataArray, según el orden del diccionario. En caso de que la cantidad de dimensiones no coincida, se asignarán las dimensiones existentes en ambos DataArrays. 

````python
# Creación del Dataset
clim_ds = xr.Dataset(data_vars={"Temperature": temp_da, "Pressure": press_da})
clim_ds
````

### Acceso a las variables y coordenadas del Dataset 

````python
# Acceso a los datos de temperatura
clim_ds.Temperature
````

````python
# Acceso a los datos de presión
clim_ds['Pressure']
````

````python
# Acceso la coordenada tiempo
clim_ds.time
````
