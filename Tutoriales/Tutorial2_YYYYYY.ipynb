Nearest-neighbor Sampling

Suppose we want to know the temperature from 2018-01-07. However, the last day on our time axis is 2018-01-05. We can therefore sample within two days of our desired date of 2018-01-07. We can do this using the .sel() method we used earlier but with the added flexibility of performing nearest neighbor sampling and specifying an optional tolerance. This is called an inexact lookup because we are not searching for a perfect match, although there may be one. Here the tolerance is the maximum distance away from our desired point Xarray will search for the nearest neighbor.

Rather than using a NumPy-like index selection, in Xarray, we can instead select data based on coordinate values using the .sel() method, which takes one or more named coordinate(s) as a keyword argument:

```python
lon_sel, lat_sel = -41, -25

sst_spatial_sel = sst_corr.sel(lon = lon_sel, lat = lat_sel, method = 'nearest')

sst_spatial_sel.sst.plot(figsize=(15, 4))
```

Operaciones de agregación espacial

```python
sst_30y = ds_sst.sel(time=slice('1980-01-01', '2010-01-01')) # Seleccionamos 30 años

sst_30y
#sst_mean = ds_sst
```

```python
# Valor medio espacial
sst_30y_mean = sst_30y.mean(dim='time')

sst_30y_mean
```

```python
fig, ax = plt.subplots(1, 3, figsize=(14, 4), sharey = True, tight_layout=True)

sst_30y_mean = sst_30y.mean(dim='time')
sst_30y_mean.sst.plot(ax=ax[0])
ax[0].set_title('Valor promedio SST (1980-2010)')

sst_30y.std(dim='time')['sst'].plot(ax=ax[1])
ax[1].set_title('Desvío estándar SST (1980-2010)')

sst_30y.max(dim='time').sst.plot(ax=ax[2])
ax[2].set_title('Valor máximo SST (1980-2010)')
```

cálculo de anomalíaS

```python
date_01 = '1985-05-27'
date_02 = '1995-05-27'
date_03 = '2005-05-27'

ds_date01 = sst_30y.sel(time=date_01)
ds_date02 = sst_30y.sel(time=date_02)
ds_date03 = sst_30y.sel(time=date_03)

ds_clim_1980to2010 = sst_30y_mean.copy()

anom_01 = ds_date01.sst - sst_30y_mean.sst
anom_02 = ds_date02.sst - sst_30y_mean.sst
anom_03 = ds_date03.sst - sst_30y_mean.sst
```

```python
fig, ax = plt.subplots(1, 3, figsize=(14, 4), sharey = True, tight_layout=True)

anom_01.plot(ax=ax[0], vmin=-1.5, vmax=2, cmap='RdBu_r')
ax[0].set_title(f'Anomalía SST {date_01}')

anom_02.plot(ax=ax[1], vmin=-1.5, vmax=2, cmap='RdBu_r')
ax[1].set_title(f'Anomalía SST {date_02}')

anom_03.plot(ax=ax[2], vmin=-1.5, vmax=2, cmap='RdBu_r')
ax[2].set_title(f'Anomalía SST {date_03}')
```
